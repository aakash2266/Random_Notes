```python
import pandas as pd
```

```python
df = pd.read_csv("sentiment.csv")
```

```python
df
```

```python
df = df[["text","sentiment"]]
```

```python
df.dropna(inplace=True)
```

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from string import punctuation
```

```python
tokens = []

for doc in df['text']:
    token = word_tokenize(doc, language = 'english', preserve_line = True)
    token = [i for i in token if i not in punctuation]
    tokens.append(token)

res = []
for doc in tokens:
    token = [i for i in doc if i not in stopwords.words('english')]
    res.append(token)
```

```python
data = {
    'sentence' : [],
    'sentiment' : [],
    }
for entry in res:
    sentence = " ".join(entry)
    data['sentence'].append(sentence)

data['sentence']
data['sentiment'] = df['sentiment']

df = pd.DataFrame(data)
df.head()
```

```python
import re
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

nltk.download("punkt")
nltk.download("stopwords")

stop_words = set(stopwords.words("english"))

def preprocess(text):
    text = re.sub(r"[^a-zA-Z0-9\s]", "", text)
    tokens = word_tokenize(text)
    tokens = [w.lower() for w in tokens if w.lower() not in stop_words]
    return " ".join(tokens)
```

```python
tokenizer = Tokenizer(num_words=5000, oov_token="<OOV>")

tokenizer.fit_on_texts(
    df["sentence"].astype(str).apply(preprocess)
)

sequences = tokenizer.texts_to_sequences(
    df["sentence"].astype(str).apply(preprocess)
)

X = pad_sequences(sequences, maxlen=20, padding="post")

```
